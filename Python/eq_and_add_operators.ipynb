## What happens with different objects and data types

What you know as `int`(egers), `float`(ing point numbers), `bool`(ean) and `char`(acters) are the building blocks of all the other data types in Python (strings, lists, arrays, dictionaries, etc.). In the language Python is built on, C, these are the primitive data types, and, although they are objects, they're first-class objects; raw data stored in a specific way so that the computer can understand them and classify them.

This is because functions have their own implementation of two methods (functions in objects) called `__repr__` and `__str__`. The first one is the one that tells the interpreter what to assign when you're assining a function to a variable, and the second one is the one that tells the interpreter what to print when you're printing a function. For instance, we could create an object that treats functions just like it would treat a primitive data type (like an int):

```python
class NewFunction:
    def __init__(self, func):
        self.func = func

    def __call__(self, *args, **kwargs):
        return self.func(*args, **kwargs)

    def __repr__(self):
        return "SIIIIUUUUU"

def return_siuu():
    print("SIIIIUUUUU")

function_ref = NewFunction(return_siuu)

# Call the function
function_ref()  # Output: SIIIIUUUUU

# Print the function reference
print(function_ref, "<-now ----- before->", return_siuu)  # Output: SIIIIUUUUU
```

# The `+` operator

Just like it happened with the `=` operator, you can twitch `+` to your liking. However, what's crucial in Python is that it normally returns a new object of the same type as the operands. For instance, if you add two integers, you get an integer, if you add two strings, you get a string, and so on. Let's see an example:



```python
a = 10
print("ID of a:", id(a))
b = a
print("ID of b:", id(b))
a += 1 #* Same as a = a + 1
print("ID of a after addition:", id(a), "--- Is b the same as a?", id(b) == id(a))
print("b:", b)
print("a:", a)
```

You are seeing `b` stores the value `10` and `a` stores the value `11` after adding `1` to it, even though `b` "was assigned" `a`. This is because, when doing `a += 1`, or `a = a + 1`, you created a *new* object and assigned it to `a`. This is why `b` didn't change, because `b` stores the ID of `a = 10` (the "first `a`"), not of `a = a + 1` (the "second" `a`), as you can see in the output.

If you want to digest it using an analogy, imaging `a` has entered a bakery and set an order for `10` coquitos. Then, `b` enters the store and asks for the exact same order (they're making a *reference* (using the assigning operator, as I told you earlier) to the order of `10` coquitos that `a` has). However, right before exitting the shop, `a` asks for one more coquito and so the cashier adds one coquito to its order; so the order of `a` has changed. However, `b`'s hasn't, and so `a` exits the shop with `11` coquitos and `b` exits the show with 10 coquitos.
